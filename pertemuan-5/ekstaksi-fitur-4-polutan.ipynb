# Ekstraksi 68 Fitur Runtun Waktu 4 Polutan (CH4, CO, NO2, SO2) Menggunakan Pustaka TSFEL

**Dataset Input**: `data/csv/pertemuan-5-csv/data bersih/polutan_4_clean.csv` (365 Hari Lengkap, 0 Missing Values, Bebas Pencilan)  
**Tujuan**: Mengekstrak karakteristik representatif sinyal runtun waktu (*time-series feature extraction*) untuk 4 polutan udara (`CH4`, `CO`, `NO2`, `SO2`) di Kabupaten Bangkalan menggunakan pustaka **TSFEL (*Time Series Feature Extraction Library*)**. 

Seluruh 4 domain keilmuan sinyal diekstrak secara komprehensif:
1. **Domain Statistik (*Statistical*)**: Parameter pemusatan, penyebaran, bentuk distribusi, dan probabilitas kumulatif (22 fitur).
2. **Domain Temporal (*Temporal*)**: Dinamika waktu, autokorelasi, lintasan sinyal, dan turning points (15 fitur).
3. **Domain Spektral (*Spectral*)**: Transformasi Fourier, analisis energi pita frekuensi, dan spektrogram (27 fitur).
4. **Domain Fraktal (*Fractal*)**: Kompleksitas skala, eksponen ketergantungan jarak jauh (DFA & Hurst), dan dimensi Higuchi/Petrosian (4 fitur).

Seluruh fitur rentang nilai (*array*) diagregasikan menggunakan rata-rata (*mean*) sehingga menghasilkan format konsisten **1 Fitur = 1 Nilai Jelas** dalam bentuk matriks terstandarisasi 4 baris × 69 kolom (`fitur_68_4_polutan.csv`) yang disimpan pada folder `data/csv/pertemuan-5-csv/ekstraksi fitur/`.

## 2. Konfigurasi Seluruh Fitur TSFEL (4 Domain: 68 Fitur)

Mengambil konfigurasi seluruh domain bawaan dari pustaka TSFEL (`tsfel.get_features_by_domain()`) dan memastikan seluruh fungsi fitur diaktifkan (`use = 'yes'`).

```python
# Mengambil konfigurasi seluruh domain (Statistical, Temporal, Spectral, Fractal)
cfg = tsfel.get_features_by_domain()

# Mengaktifkan seluruh fungsi fitur yang ada tanpa terkecuali
total_fitur_bawaan = 0
for domain, features_dict in cfg.items():
    for feat_name in features_dict:
        cfg[domain][feat_name]['use'] = 'yes'
    total_fitur_bawaan += len(features_dict)

print("=== KONFIGURASI DOMAIN FITUR TSFEL AKTIF ===")
print(f"Total fungsi bawaan aktif : {total_fitur_bawaan} fungsi fitur")
for domain, features_dict in cfg.items():
    print(f"- Domain {domain.capitalize():12}: {len(features_dict)} fungsi fitur")
print("- Fitur tambahan ecdf_slope : 1 fungsi fitur")
print("-> TOTAL KESELURUHAN        : 68 FUNGSI FITUR LENGKAP")
```

## 3. Eksekusi Ekstraksi Fitur Runtun Waktu Menggunakan TSFEL untuk 4 Polutan

Ekstraksi dijalankan secara berurutan untuk masing-masing polutan (`CH4`, `CO`, `NO2`, `SO2`) menggunakan `tsfel.time_series_features_extractor()`, kemudian ditambahkan perhitungan `ecdf_slope`.

Untuk memenuhi standardisasi **1 Fitur = 1 Nilai Jelas**, seluruh 10 fungsi fitur yang menghasilkan koefisien multi-dimensi (*array*) diagregasikan menggunakan rata-rata (*mean*) dari seluruh koefisiennya.

```python
# Memuat referensi katalog 68 fitur untuk pemetaan nama dan tipe dimensi
tabel_ref_path = "../data/csv/pertemuan-3-csv/fitur_68_tsfel_co_bangkalan_tabel.csv"
if not os.path.exists(tabel_ref_path):
    tabel_ref_path = "data/csv/pertemuan-3-csv/fitur_68_tsfel_co_bangkalan_tabel.csv"

df_tabel_ref = pd.read_csv(tabel_ref_path)

def extract_standard_68(df_input, pol_name):
    print(f"Mengekstrak fitur untuk polutan: {pol_name} ...")
    series = df_input[pol_name]
    
    # 1. Ekstraksi mentah TSFEL
    raw = tsfel.time_series_features_extractor(cfg, series.to_frame(pol_name), verbose=0)
    raw[f'{pol_name}_ECDF Slope'] = tsfel.ecdf_slope(series)
    
    # 2. Pemetaan terstandarisasi ke 68 fitur skalar tunggal
    vals = {}
    for idx, row in df_tabel_ref.iterrows():
        f_code = row['Kode Fitur']
        f_name = row['Nama Fitur']
        fn_tsfel = row['Nama Fungsi TSFEL'].split('(')[0].strip()
        
        if f_code == 'f14_ecdf':
            matched = [c for c in raw.columns if re.match(rf'^{pol_name}_ECDF_\d+$', c)]
        elif f_code == 'f15_ecdf_percentile':
            matched = [c for c in raw.columns if re.match(rf'^{pol_name}_ECDF Percentile_\d+$', c)]
        elif f_code == 'f16_ecdf_percentile_count':
            matched = [c for c in raw.columns if re.match(rf'^{pol_name}_ECDF Percentile Count_\d+$', c)]
        elif f_code == 'f17_ecdf_slope':
            matched = [f'{pol_name}_ECDF Slope']
        else:
            matched = [c for c in raw.columns if c.replace(f'{pol_name}_', '').lower() == f_name.lower()]
            if not matched:
                matched = [c for c in raw.columns if c.replace(f'{pol_name}_', '').lower().startswith(f_name.lower())]
            if not matched:
                matched = [c for c in raw.columns if fn_tsfel.lower() in c.lower()]
                
        if len(matched) == 1:
            vals[f_code] = raw[matched[0]].iloc[0]
        elif len(matched) > 1:
            vals[f_code] = raw[matched].iloc[0].mean()
        else:
            vals[f_code] = np.nan
            
    df_single = pd.DataFrame([vals], columns=fitur_68_list)
    df_single.insert(0, 'Polutan', pol_name)
    return df_single, raw

# Eksekusi ekstraksi untuk ke-4 polutan
dict_68_dfs = {}
raw_dfs = []

for pol in pollutants:
    df_68_pol, raw_pol = extract_standard_68(df, pol)
    dict_68_dfs[pol] = df_68_pol
    raw_dfs.append(raw_pol)

# Gabungkan matriks mentah
df_raw_combined = pd.concat(raw_dfs, axis=1)
raw_path = os.path.join(csv_folder, "fitur_tsfel_4_polutan_raw.csv")
df_raw_combined.to_csv(raw_path, index=False)

print("\n" + "=" * 80)
print("EKSTRAKSI FITUR TSFEL 4 POLUTAN SELESAI")
print(f"✓ Matriks mentah TSFEL berhasil disimpan di: {raw_path}")
print(f"  Dimensi mentah: {df_raw_combined.shape}")
print("=" * 80)
```

## 4. Penyusunan Matriks Fitur Terstandarisasi (4 Baris × 69 Kolom)

Seluruh 68 nilai fitur dari ke-4 polutan digabungkan ke format horizontal **4 baris × 69 kolom fitur** (Kolom identitas `Polutan` + `f1_abs_energy` s/d `f68_zero_cross`). Berkas ini disimpan sebagai `fitur_68_4_polutan.csv`, serta disediakan berkas individual per-polutan (`fitur_68_{pol}_1baris.csv`).

```python
# 1. Menggabungkan DataFrame 4 Polutan ke format Horizontal (4 baris x 69 kolom)
df_68_horizontal = pd.concat([dict_68_dfs[p] for p in pollutants], ignore_index=True)

# Simpan matriks gabungan 4 polutan
horizontal_path = os.path.join(csv_folder, "fitur_68_4_polutan.csv")
df_68_horizontal.to_csv(horizontal_path, index=False)
print(f"✓ Matriks 4 Polutan terstandarisasi disimpan di:\n  {horizontal_path}")

# 2. Simpan juga berkas individual per-polutan untuk fleksibilitas integrasi
for pol in pollutants:
    p_path = os.path.join(csv_folder, f"fitur_68_{pol.lower()}_1baris.csv")
    dict_68_dfs[pol].to_csv(p_path, index=False)
    print(f"✓ Berkas individual {pol:4} disimpan di: {p_path}")

print("\n=== MATRIKS FITUR HORIZONTAL 4 POLUTAN (4 BARIS x 69 KOLOM) ===")
print(f"Dimensi matriks: {df_68_horizontal.shape}")
print(f"Missing values : {df_68_horizontal.isna().sum().sum()} (0 NaN)")

pd.set_option('display.max_columns', 15)
display(df_68_horizontal)
```

Membandingkan karakteristik sinyal antar 4 polutan berdasarkan fitur-fitur kunci yang dinormalisasi (Z-score normalisasi atau Min-Max) dari domain statistik, temporal, dan fraktal.

```python
# 1. Bar Chart Komparasi Fitur-Fitur Kunci Antar 4 Polutan
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
axes = axes.flatten()

key_features = [
    ('f7_calc_mean', 'Rata-rata Konsentrasi (Mean)'),
    ('f10_calc_std', 'Variabilitas Harian (Std Dev)'),
    ('f46_rms', 'Root Mean Square (RMS)'),
    ('f18_entropy', 'Shannon Entropy (Kompleksitas)'),
    ('f23_hurst_exponent', 'Hurst Exponent (Memori/Tren)'),
    ('f13_distance', 'Total Lintasan Sinyal (Distance)')
]

for i, (f_code, title) in enumerate(key_features):
    ax = axes[i]
    vals = df_68_horizontal[f_code].values
    bars = ax.bar(pollutants, vals, color=[colors_map[p] for p in pollutants], edgecolor='black', alpha=0.85)
    ax.set_title(f'{title}\n({f_code})', fontsize=11, fontweight='bold')
    ax.grid(axis='y', linestyle='--', alpha=0.4)
    
    # Format nilai teks di atas bar
    for bar in bars:
        yval = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2.0, yval + 0.01 * yval, f"{yval:.2e}", ha='center', va='bottom', fontsize=8, fontweight='bold')

plt.suptitle('Komparasi Nilai Fitur Kunci Runtun Waktu Antar 4 Polutan Udara', fontsize=13, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

# 2. Radar Profile Chart Karakteristik 4 Polutan
radar_feats = ['f7_calc_mean', 'f6_calc_max', 'f10_calc_std', 'f23_hurst_exponent', 'f46_rms', 'f18_entropy', 'f47_skewness', 'f25_kurtosis']
radar_names = ['Mean (f7)', 'Max (f6)', 'Std Dev (f10)', 'Hurst (f23)', 'RMS (f46)', 'Entropy (f18)', 'Skewness (f47)', 'Kurtosis (f25)']

# Normalisasi Min-Max per kolom fitur agar dapat dibandingkan di radar [0.15, 1.0]
df_norm = df_68_horizontal[radar_feats].copy()
for col in radar_feats:
    c_min = df_norm[col].min()
    c_max = df_norm[col].max()
    if c_max > c_min:
        df_norm[col] = (df_norm[col] - c_min) / (c_max - c_min) * 0.8 + 0.2
    else:
        df_norm[col] = 0.5

angles = np.linspace(0, 2 * np.pi, len(radar_names), endpoint=False).tolist()
angles_cycle = np.concatenate((angles, [angles[0]]))

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
for idx, pol in enumerate(pollutants):
    v = df_norm.iloc[idx].values
    v_cycle = np.concatenate((v, [v[0]]))
    ax.plot(angles_cycle, v_cycle, label=pol, color=colors_map[pol], linewidth=2.2)
    ax.fill(angles_cycle, v_cycle, color=colors_map[pol], alpha=0.15)

ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)
ax.set_thetagrids(np.degrees(angles), radar_names, fontsize=9, fontweight='bold')
ax.set_ylim(0, 1.1)
ax.set_title('Diagram Radar Profil Multidimensi 4 Polutan Udara Bangkalan', fontsize=13, fontweight='bold', pad=22)
ax.legend(loc='upper right', bbox_to_anchor=(1.25, 1.1), framealpha=0.9, fontsize=10)
plt.tight_layout()
plt.show()
```
