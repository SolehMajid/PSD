```python
import folium

aoi = {
  "type": "FeatureCollection",
  "features": [
    {
      "type": "Feature",
      "properties": {},
      "geometry": {
        "type": "Polygon",
        "coordinates": [
          [
            [112.6736254, -7.063601],
            [112.6736254, -7.075242],
            [112.6865832, -7.0936504],
            [112.7010414, -7.0960868],
            [112.7059517, -7.1058321],
            [112.7018597, -7.1196376],
            [112.7283209, -7.1242393],
            [112.7446886, -7.1155772],
            [112.7479622, -7.1055614],
            [112.7738777, -7.1131409],
            [112.7899726, -7.1109754],
            [112.7866991, -7.0841756],
            [112.7648755, -7.0871534],
            [112.7689674, -7.077137],
            [112.7528725, -7.0749713],
            [112.744143, -7.0592693],
            [112.7062245, -7.0525011],
            [112.6759442, -7.0522304],
            [112.6739496, -7.0571792],
            [112.6736254, -7.063601]
          ]
        ]
      }
    }
  ]
}

# Titik tengah AOI (Kecamatan Socah, Bangkalan)
coords = aoi["features"][0]["geometry"]["coordinates"][0]
center_lat = sum(c[1] for c in coords) / len(coords)
center_lon = sum(c[0] for c in coords) / len(coords)

# Buat peta
m = folium.Map(
    location=[center_lat, center_lon],
    zoom_start=12,
    tiles="OpenStreetMap"
)

# Tambahkan AOI
folium.GeoJson(
    aoi,
    name="AOI Socah",
    style_function=lambda feature: {
        "fillColor": "blue",
        "color": "red",
        "weight": 2,
        "fillOpacity": 0.3
    }
).add_to(m)

# Tambahkan kontrol layer
folium.LayerControl().add_to(m)

# Tampilkan
m
```
