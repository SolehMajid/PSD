```python
import pandas as pd
import tsfel

# 1. Baca file Excel
df = pd.read_excel("in class.xlsx")

# 2. Ambil hanya kolom sinyal yang berisi angka ('Nilai')
signal_data = df[["Nilai"]]

# 3. Konfigurasi TSFEL khusus untuk fitur RMS
cfg = {
    "statistical": {
        "Root mean square": {
            "Description": "Calculates root mean square.",
            "parameters": {},
            "function": "tsfel.rms",
            "use": "yes",
        }
    }
}

# 4. Ekstrak fitur RMS
df_rms = tsfel.time_series_features_extractor(cfg, signal_data)

# 5. Tampilkan hasil
print(df_rms)
```
