```python
import folium

aoi = {
    "type": "FeatureCollection",
    "features": [
        {
            "type": "Feature",
            "properties": {},
            "geometry": {
                "type": "Polygon",
                "coordinates": [
                    [
                        [
              112.6654991,
              -7.0171347
            ],
            [
              112.6912024,
              -7.1679391
            ],
            [
              112.7995564,
              -7.0918788
            ],
            [
              112.6993392,
              -7.0334062
            ],
            [
              112.6654991,
              -7.0171347
            ]
                    ]
                ]
            }
        }
    ]
}

# Titik tengah AOI
center_lat = (-7.1613419 + -7.2296572 + -6.8602645 + -6.8602645) / 4
center_lon = (112.6669403 + 113.0865997 + 113.0865997 + 112.6725233) / 4

# Buat peta
m = folium.Map(
    location=[center_lat, center_lon],
    zoom_start=10,
    tiles="OpenStreetMap"
)

# Tambahkan AOI
folium.GeoJson(
    aoi,
    name="AOI",
    style_function=lambda feature: {
        "fillColor": "blue",
        "color": "red",
        "weight": 2,
        "fillOpacity": 0.3
    }
).add_to(m)

# Tambahkan kontrol layer
folium.LayerControl().add_to(m)

# Tampilkan
m
```
